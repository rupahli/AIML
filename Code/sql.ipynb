```python
import pandas as pd

# data taken from https://www.kaggle.com/datasets/ajitashwath/gaming-addiction-dataset
df= pd.read_csv(r"C:\Users\KITTU-PALKIN\OneDrive\Documents\Rupahli\GUVI\AIML\Data\gaming_addiction.csv") # -- Reading a CSV file into a DataFrame



dtype_mapping ={
    'int64':"INT",
    'float64':"FLOAT",
    'str':"varchar(100)",
    'bool':"BOOLEAN",
    'object':"varchar(100)"
}
columns=[]
for col, dtype in df.dtypes.items():
    #print (col, dtype)
    dtype = dtype_mapping.get(str(dtype),"varchar(255)") ## default varchar if not found in mapping
    columns.append(f"{col} {dtype}")

columns

query = "create table gaming (" + (',').join(columns) +")"

cursor.execute(query)

query = "insert into gaming (" + (',').join(columns) +")"

cursor.execute(query)
```
