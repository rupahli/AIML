```python
import pandas as pd

df= pd.read_csv(r"C:\Users\KITTU-PALKIN\OneDrive\Documents\Rupahli\GUVI\AIML\Data\Bengaluru_House_Data.csv") 
df.drop_duplicates(inplace=True)
condition = df.isna().sum()*100/len(df) <= 30 # which columns have less than 30 % null values
selected_cols = df.columns[condition]
df = df[selected_cols]
#df.describe()
#df.info()

df['total_sqft'] = pd.to_numeric(df['total_sqft'],errors="coerce")
df.isna().sum()
df['total_sqft'] = df['total_sqft'].fillna(df['total_sqft'].mean()) # we have filled na values with mean 
df.isna().sum()
df['bath'] = df['bath'].fillna(df['bath'].median()) # we have filled na values with mean 
df.isna().sum()
df['balcony'] = df['balcony'].fillna(df['balcony'].median()) # we have filled na values with mean 
df.isna().sum()
df['location'] = df['location'].fillna(df['location'].mode()[0]) # [0] because in case 2 value are hightest then take first one
df.isna().sum()
df['size'] = df['size'].fillna(df['size'].mode()[0]) # [0] because in case 2 value are hightest then take first one
df.isna().sum()

```
