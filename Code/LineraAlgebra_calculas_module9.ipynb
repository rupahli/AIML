```python
import numpy as np

np.random.randint(1,10) #--- Whole numbers (int)
np.random.rand() # generate one random number between 0 and 1 ---Floating-point decimals (float)
np.random.rand(5) # generate 5 random numbers between 0 and 1 ---Floating-point decimals (float)


np.random.seed(42)
np.random.rand(5) # now they will be same every time you run the code because of the seed value
```

```python
coin = np.random.choice(['H','T'],size=10) # generate 10 random coin tosses
coin
```

```python
coin = np.random.choice(['H','T'],size=20) # generate 10 random coin tosses
coin

probability_heads= 0
probability_tails= 0

for p in coin:
    if p == 'H':
        probability_heads += 1
    else:
        probability_tails += 1
        

ph = probability_heads / len(coin)
pt = probability_tails / len(coin)

print(ph, pt)
```

```python
import numpy as np
import matplotlib.pyplot as plt

coin = np.random.choice(['H','T'],size=20000) # generate 20000 random coin tosses -- higher the number of tosses, 
#the more accurate the probability will be
coin

head = list(coin).count('H') # count the number of heads
tail = 20000-head

plt.bar(['Heads','Tails'],[head,tail])
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Normal Distribution

data = np.random.normal(loc=10, scale=1.4, size=20) 

data
# generate 1000 random numbers from a normal distribution with mean 10 and standard deviation 1.4

plt.hist(data, bins=10) # plot a histogram with 10 bins
plt.show()

# we should increase samples

data = np.random.normal(loc=10, scale=1.4, size=200000) 

print(data)
# generate 1000 random numbers from a normal distribution with mean 10 and standard deviation 1.4

plt.hist(data, bins=20) # plot a histogram with 10 bins
plt.show() # this is better than the previous one because we have more samples and more bins
```

```python
import numpy as np
import matplotlib.pyplot as plt

# UNIFORM Distribution - no change in data distribution, all values are equally likely to occur

# we should increase samples

data = np.random.uniform(low=0, high=500, size=200000) 
# uniform distribution generates random numbers between a specified range (low and high) with equal probability

data
# generate 1000 random numbers from a normal distribution with mean 10 and standard deviation 1.4

plt.hist(data, bins=20) # plot a histogram with 10 bins
plt.show() # this is bett
```

```python
#Poisson Distribution is a Discrete Distribution. It estimates how many times an event can happen in a specified time.
data = np.random.poisson(lam=5, size=2000000)
## Generate 2000000 random integers simulating traffic (average rate = 5 events)
plt.hist(data, bins=20) 
plt.show() 

```

```python
# exponential Distribution is a continuous probability distribution that describes the time between events in a Poisson
#  point process. It is often used to model the time until an event occurs, such as the time until a radioactive
#  particle decays or the time until a customer arrives at a service center.

# how many calls are received by a call center in a given time period -- poisson
# how much time we have to wait for the next call -- exponential

data=np.random.exponential(scale=1.0, size=2000000)
plt.hist(data, bins=20)
plt.show()
```
