```python
#nominal  - we can not rank our data --> one hot encoding 

# ordinal - we can rank --> ordinal encoding

# binary - label encoding


# OneHotEncoder and pd.get_dummies() both perform one-hot encoding, but they serve completely different stages of a machine learning 
# project. OneHotEncoder is built for production workflows and machine learning pipelines, while pd.get_dummies() is designed 
# for fast, exploratory data analysis (EDA).

# imagine you train a model on a dataset containing three cities: London, Paris, and Tokyo. 
# 
# 1. The pd.get_dummies() FailureIf your deployment API receives a single test production request for Paris, pd.get_dummies() will 
# generate exactly one column: # [City_Paris]. Your trained model, which expects three columns (City_London, City_Paris, City_Tokyo), will instantly crash due 
# to a structural shape mismatch.If the API receives an unknown city like New York, pd.get_dummies() creates a brand-new column 
# [City_New York]. Your model has no idea what to do with this new dimension.
# 
# 2. The OneHotEncoder SolutionOneHotEncoder memorizes 
# the exact columns from your training set.If it processes a lone data point for Paris, it still outputs all three original columns, 
# filling Paris with a 1 and the rest with 0.If it encounters New York, setting handle_unknown='ignore' forces it to safely output 
# [0, 0, 0], matching the expected layout perfectly without breaking your pipeline.
```

```python
import pandas as pd
from sklearn.preprocessing import OrdinalEncoder

df = pd.read_csv(r"C:\Users\KITTU-PALKIN\OneDrive\Documents\Rupahli\GUVI\AIML\Data\fifa_world_cup_2026_player_performance.csv")

df.drop(['player_id','preferred_foot','player_name','match_result','decelerations','nationality','pressure_resistance','possession_impact','defensive_contribution','offensive_contribution','goals_team','goals_opponent','team','jersey_number','position','club_name','market_value_eur','match_id','match_date','stadium','city','creativity_score','consistency_score','clutch_performance_score','total_goals_tournament','total_assists_tournament','tournament_rating'],axis=1,inplace=True)
# Nominal
#.       we cannot rank our data--> One hot encoding
# Ordinal
#.       tournament_stage--> Ordinal Encoding
# Binary 
#.       Label Encoding

# OneHotEncoder and pd.get_dummies() both perform one-hot encoding, but they serve completely different stages of a machine learning 
# project. OneHotEncoder is built for production workflows and machine learning pipelines, while pd.get_dummies() is designed 
# for fast, exploratory data analysis (EDA).

# imagine you train a model on a dataset containing three cities: London, Paris, and Tokyo. 
# 
# 1. The pd.get_dummies() FailureIf your deployment API receives a single test production request for Paris, pd.get_dummies() will 
# generate exactly one column: # [City_Paris]. Your trained model, which expects three columns (City_London, City_Paris, City_Tokyo), will instantly crash due 
# to a structural shape mismatch.If the API receives an unknown city like New York, pd.get_dummies() creates a brand-new column 
# [City_New York]. Your model has no idea what to do with this new dimension.
# 
# 2. The OneHotEncoder SolutionOneHotEncoder memorizes 
# the exact columns from your training set.If it processes a lone data point for Paris, it still outputs all three original columns, 
# filling Paris with a 1 and the rest with 0.If it encounters New York, setting handle_unknown='ignore' forces it to safely output 
# [0, 0, 0], matching the expected layout perfectly without breaking your pipeline.


df = pd.get_dummies(df,columns=['opponent_team'],dtype='int',drop_first=True) #Nominal # OneHotEncoder and pd.get_dummies() both perform one-hot encoding
encoder = OrdinalEncoder(categories=[[ "Group Stage", #Ordinal
        "Round of 32",
        "Round of 16",
        "Quarter Finals",
        "Semi Finals",
        "Third Place Match",
        "Final"]])

df['tournament_stage'] = encoder.fit_transform(df[['tournament_stage']])
for col in df.columns:
    print(col)

cat_col = df.select_dtypes(include='object').columns.to_list() # to apply above nominal or ordinal data if any remaining
cat_col # nothing is there hence no more encoding required 
df.info()

df.head()

df['tournament_stage']

print(df.tail(10))
```
