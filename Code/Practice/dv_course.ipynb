```python
from datetime import datetime, timedelta
import matplotlib.pyplot as plt
import pandas as pd
import random

# 1. Generate sample time series data
start_date = datetime(2026, 1, 1)
dates = [start_date + timedelta(days=i) for i in range(60)]
values = [100 + (i * 1.5) + random.uniform(-10, 10) for i in range(60)]

# 2. Build a DataFrame (Best practice for time series data)
df = pd.DataFrame({"Date": dates, "Value": values})

# 3. Initialize the plot
plt.figure(figsize=(10, 5))
plt.plot(df["Date"], df["Value"], color="tab:blue", linewidth=2, label="Daily Metrics")

# 4. Format the presentation
plt.title("Metrics Tracking Over Time", fontsize=14, fontweight="bold", pad=15)
plt.xlabel("Date", fontsize=12)
plt.ylabel("Value (Units)", fontsize=12)
plt.grid(True, linestyle="--", alpha=0.6)
plt.legend()

plt.axvline(pd.Timestamp("2026-01-06"))
plt.axhspan(120,140,color="green",alpha=0.5)

# 5. Automatically adjust date labels to prevent overlap
plt.gcf().autofmt_xdate()

# 6. Render the graph
plt.show()
```
