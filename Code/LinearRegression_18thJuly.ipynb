# Cross validation ----is a resampling statistical technique used to evaluate the performance of machine learning models on unseen data while actively preventing overfitting

# K-Fold Cross-Validation: Splits the full dataset into K equal-sized subsets (folds). The model trains on K-1 folds and tests on the remaining single fold. This entire loop repeats K times so each fold serves as the test set exactly once. The final score is the average performance across all K iterations.

```python
import pandas as pd
import seaborn as sns

df = pd.read_csv(r"C:\Users\KITTU-PALKIN\OneDrive\Documents\Rupahli\GUVI\AIML\Data\extended_flower_morphometrics.csv")

X = df.drop(["species"],axis = 1)
Y = df["species"]


from sklearn.model_selection import train_test_split


x_train,x_test , y_train,y_test = train_test_split(X,Y,test_size = 0.2)

from sklearn.linear_model import LogisticRegressionCV

model = LogisticRegressionCV(cv=7)

model.fit(x_train,y_train)


train_pred = model.predict(x_train)

test_pred = model.predict(x_test)

from sklearn.metrics import accuracy_score,precision_score,recall_score, f1_score

print(f"accuracy_score on training data : {accuracy_score(y_train,train_pred)}")
print(f"precision_score on training data : {precision_score(y_train,train_pred,average="micro")}")
print(f"recall_score on training data : {recall_score(y_train,train_pred,average="micro")}")
print(f"f1_score on training data : {f1_score(y_train,train_pred,average="micro")}")

print("\n")

print(f"accuracy_score on test data : {accuracy_score(y_test,test_pred)}")
print(f"precision_score on test data : {precision_score(y_test,test_pred,average="micro")}")
print(f"recall_score on test data : {recall_score(y_test,test_pred,average="micro")}")
print(f"f1_score  on test data : {f1_score(y_test,test_pred,average="micro")}")

```

```python
from yellowbrick.cluster import KElbowVisualizer
import pandas as pd
from sklearn.preprocessing import LabelEncoder
from sklearn.cluster import KMeans
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\KITTU-PALKIN\OneDrive\Documents\Rupahli\GUVI\AIML\Data\Mall_Customers.csv")
df.drop(['CustomerID'],axis=1,inplace=True)
encoder = LabelEncoder()
df['Gender'] = encoder.fit_transform(df['Gender'])
model = KMeans()

visualization = KElbowVisualizer(model,k = (2,10), force_model=True)
visualization.fit(df)
visualization.show()

## As my model shows 5 clusters so we will use n_clusters=5 in KMeans and visualize the clusters

model = KMeans(n_clusters=5)
model.fit(df)

# to check accurcy we will use silhouette score -- which is a metric to check how well the clusters are formed

from sklearn.metrics import silhouette_score

print(f"Sillhouette Score: {silhouette_score(df, model.labels_)}")
#Sillhouette Score: 0.42672573382077056 -- this should be as close to 1 as possible -- so this is good score
# as our score is 0.42 so our model is not that good

# 1 means cluster data points are far away
# 0 means cluster data points are close to each other - touching boundries
# -1 means cluster data points are overlapping each other -- wrong clusters


df["Label"] = model.labels_

# After clustering -- same data points with three diff colors divided in three clusters
sns.scatterplot(df,x="Annual Income (k$)",y="Spending Score (1-100)",hue="Label",palette='Set2')
plt.show()
```
