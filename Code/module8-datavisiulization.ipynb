```python
import pandas as pd
import numpy as np

##Recommended for large datasets where scatter plots become dense, effectively visualizing data density. Each hexagon's color 
# intensity (darker # shades indicating higher density) helps reveal concentrations in data, particularly 
# useful for geospatial or customer segmentation

data = {
'Age': np.random.randint(20,70,500),
'Salary': np.random.randint(50000,300000,500),
'Experience': np.random.randint(1,30,500),
'Performance': np.random.randint(40,100,500),
'Department': np.random.choice(['AIML','DS','DA'],500),
'Gender': np.random.choice(['Female','Male'],500)
}


df = pd.DataFrame(data)

plt.hexbin(df['Age'],df['Salary'],gridsize=20,cmap='Blues') # hexbin is used for 2D data
plt.xlabel("Age")
plt.ylabel("Salary")
plt.show()
```

```python
plt.scatter(df['Age'],df['Salary'],s=df['Experience'])
plt.show()
```

```python
plt.scatter(df['Age'],df['Salary'],s=df['Experience'],c=df['Performance'],cmap='viridis') # cmap is used for color map
plt.show()
```

```python
dept_salary = df.groupby(['Department'])['Salary'].mean()
dept_salary
## 2.	Lollipop Graphs: Presented as an intuitive, aesthetic alternative to traditional bar charts, using plt.hlines 
# for lines and plt.plot for "dots" to represent categorical values.

plt.hlines(y=dept_salary.index,xmin=0,xmax=dept_salary.values)
plt.plot(dept_salary.values,dept_salary.index,'*')
plt.show()

dept_salary.index #Index(['AIML', 'DA', 'DS'], dtype='object', name='Department') -- first column as index
dept_salary.values #array([173566.35087719, 178807.40236686, 180559.6       ])
```

```python
##9.	Error Bar Graphs: Highlight data differences and dispersion, typically showing mean values 
# with vertical lines representing standard deviation, common in clinical trials and experimental data.


import pandas as pd
import numpy as np

data = {
'Age': np.random.randint(20,70,500),
'Salary': np.random.randint(50000,300000,500),
'Experience': np.random.randint(1,30,500),
'Performance': np.random.randint(40,100,500),
'Department': np.random.choice(['AIML','DS','DA'],500),
'Gender': np.random.choice(['Female','Male'],500)
}

#9.	Error Bar Graphs: Highlight data differences and dispersion, typically showing
#  mean values with vertical lines representing standard deviation, common in clinical trials and experimental data.


df = pd.DataFrame(data)

dept = df.groupby(['Department'])

mean = dept.Performance.mean()  
std = dept.Performance.std()

std 
#Department
#AIML    17.601989
#DA      17.375510
#DS      17.298283

mean
#Department
#AIML    68.070513
#DA      68.773256
#DS      68.924419

plt.errorbar(mean.index,mean,yerr=std)
plt.show()

```

```python
## timeseries graph


import pandas as pd
import numpy as np

data = {
'Date': pd.date_range(start='2025-01-01', periods=100),
'Sales':np.random.randint(200,400,100)
}

df = pd.DataFrame(data)

df['MV'] = df['Sales'].rolling(window=7).mean() # moving average -- rolling helps to calculate moving average .. 
#after every 7 window size (days .. or whatever date column is having after every7 interval) it will calculate 
# the average of salary for those 7 records and store in new column MV

plt.plot(df['Date'],df['Sales'],label='Sales')
plt.plot(df['Date'],df['MV'],label='Moving Average')
plt.legend()
plt.grid() # grid lines
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns

data = {
'Age': np.random.randint(20,70,500),
'Salary': np.random.randint(50000,300000,500),
'Experience': np.random.randint(1,30,500),
'Performance': np.random.randint(40,100,500),
'Department': np.random.choice(['AIML','DS','DA'],500),
'Gender': np.random.choice(['Female','Male'],500)
}

#9.	violine graph - it helps in visualizing the distribution of the data across different categories, 
# combining aspects of box plots and density plots to show the probability density of the data at different values.


df = pd.DataFrame(data)
sns.violinplot(x='Department',y='Salary',data=df)
plt.show()

sns.swarmplot(x='Department',y='Salary',data=df) # in dots format
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

data = {
'Age': np.random.randint(20,70,500),
'Salary': np.random.randint(50000,300000,500),
'Experience': np.random.randint(1,30,500),
'Performance': np.random.randint(40,100,500),
'Department': np.random.choice(['AIML','DS','DA'],500),
'Gender': np.random.choice(['Female','Male'],500)
}

#9.	violine graph - it helps in visualizing the distribution of the data across different categories, 
# combining aspects of box plots and density plots to show the probability density of the data at different values.


df = pd.DataFrame(data)
sns.violinplot(x='Department',y='Salary',data=df)
#plt.show()

sns.swarmplot(x='Department',y='Salary',data=df) # in dots format
#plt.show()

plt.savefig("violin_swarm_plot.png",dpi = 300) # to save the plot as an image file
plt.savefig("violin_swarm_plot.pdf") # to save the plot as an image file
plt.savefig("violin_swarm_plot.svg") # to save the plot as an image file
```
